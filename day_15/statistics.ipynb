## TASK 01__Descriptive statistics on a real dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats


# Load tha dataset
student_data = pd.read_csv("student_performance (3).csv")

# Remove missing values
student_data = student_data.dropna()

# Remove the impossible 99-hour study outlier
student_data = student_data[student_data["study_hours"] != 99]


# Select the columns for analysis
final_scores = student_data["final_score"]
study_hours = student_data["study_hours"]


# Statistics for Final score

final_score_mean = final_scores.mean()
final_score_median = np.median(final_scores)
final_score_mode = stats.mode(final_scores, keepdims=True).mode[0]
final_score_standard_deviation = final_scores.std(ddof=1)
final_score_variance = final_scores.var(ddof=1)
final_score_percentile = np.percentile(final_scores,[25, 50, 75])


print("FINAL SCORE STATISTICS")
print("Mean:", final_score_mean)
print("Median:",final_score_median)
print("Mode:", final_score_mode)
print("Standard Deviation:", final_score_standard_deviation)
print("Variance:", final_score_variance)
print("25th Percentile:", final_score_percentiles[0])
print("50th Percentile:", final_score_percentiles[1])
print("75th Percentile;", final_score_percentile[2])


# Statistic for Study Hours

study_hours_mean = study_hours.mean()
study_hours_median = np.median(study_hours)
study_hours_mode = stats.mode(study_hours, keepdims=True).mode[0]
study_hours_standard_deviation = study_hours.std(ddof=1)
study_hours_variance = study_hours.var(ddof=1)
study_hours_percentiles = np.percentile(study_hours, [25, 50, 75])

print("\nSTUDY HOURS STATISTICS")
print("Mean:", study_hours_mean)
print("Median:", study_hours_median)
print("Mode:", study_hours_mode)
print("Standard Deviation:", study_hours_standard_deviation)
print("Variance:", study_hours_variance)
print("25th Percentile:", study_hours_percentiles[0])
print("50th Percentile:", study_hours_percentiles[1])
print("75th Percentile:", study_hours_percentiles[2])


# Histogram of Final Score

plt.figure(figsize=(8, 5))

plt.hist(final_scores, bins=10, edgecolor= "black")

plt.axvline(
    final_score_mean,
    linestyle="--",
    label=f"Mean = {final_score_mean:.2f}"
)

plt.xlabel("Final Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Final Scores")
plt.legend()

plt.show()



# Compare Mean and Median

print("\nMEAN VS MEDIAN")

if abs(final_score_mean - final_score_median) < 2:
    print("Mean and median are close, so the data is approximately symmetric.")
elif final_score_mean > final_score_median:
    print("Mean is higher than median, suggesting right-skewed data.")
else:
    print("Mean is lower than median, suggesting left-skewed data.")






```

1. When is the median a better summary than the mean?

   → When the data has **outliers or is highly skewed**, because the median is less affected by extreme values.

2. Why does an outlier pull the mean but not the median?

   → The mean uses **every value**, while the median depends only on the **middle position** of the sorted data.

3. What does it mean if the mean is well above the median?

   → It usually means the data is **right-skewed**, with some unusually high values.



## TASK 02__A 95% confidence interval

```python
import numpy as np
import pandas as pd
from scipy import stats

# Load your CSV file
student_data = pd.read_csv("student_performance (3).csv")

# Remove missing values
student_data = student_data.dropna()

# Remove the invalid 99-hour study value
student_data = student_data[student_data["study_hours"] != 99]

# Select final scores
final_scores = student_data["final_score"]



# 1. Mean and Standard Error

sample_size = len(final_scores)

mean_final_score = final_scores.mean()

standard_error = (
    final_scores.std(ddof=1) / np.sqrt(sample_size)
)

print("MEAN AND STANDARD ERROR")
print("Sample Size:", sample_size)
print("Mean Final Score:", mean_final_score)
print("Standard Error:", standard_error)



# 2. 95% Confidence Interval

lower_bound, upper_bound = stats.t.interval(
    0.95,
    df=sample_size - 1,
    loc=mean_final_score,
    scale=standard_error
)

print("\n95% CONFIDENCE INTERVAL")
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print(
    f"We are 95% confident that the true average "
    f"final score lies between {lower_bound:.2f} "
    f"and {upper_bound:.2f}."
)



# 3. Function to calculate confidence interval

def calculate_confidence_interval(score_data):
    sample_size = len(score_data)
    mean_score = score_data.mean()

    standard_error = (
        score_data.std(ddof=1) / np.sqrt(sample_size)
    )

    lower_bound, upper_bound = stats.t.interval(
        0.95,
        df=sample_size - 1,
        loc=mean_score,
        scale=standard_error
    )

    interval_width = upper_bound - lower_bound

    return lower_bound, upper_bound, interval_width




# 4. Pretend we have only 30 students

scores_30_students = final_scores.iloc[:30]

lower_30, upper_30, width_30 = calculate_confidence_interval(
    scores_30_students
)



# 5. Pretend we have 2000 students

scores_2000_students = final_scores.iloc[:2000]

lower_2000, upper_2000, width_2000 = calculate_confidence_interval(
    scores_2000_students
)




# 6. Compare the confidence intervals

print("\nCONFIDENCE INTERVAL COMPARISON")

print(
    f"30 students   : {lower_30:.2f} to {upper_30:.2f}"
    f" | Width = {width_30:.2f}"
)

print(
    f"2000 students : {lower_2000:.2f} to {upper_2000:.2f}"
    f" | Width = {width_2000:.2f}"
)




# 7. Meaning

print("\nEXPLANATION")

print(
    "The interval becomes narrower as the sample size increases "
    "because the standard error decreases."
)

print(
    "95% confidence means that if we repeatedly took samples "
    "and calculated confidence intervals in the same way, "
    "about 95% of those intervals would contain the true population mean."
)
```

## TASK 03__A t-test: do two groups differ?

```python
import numpy as np
import pandas as pd
from scipy import stats

# Load the CSV file
student_data = pd.read_csv("student_performance (3).csv")

# Remove missing values
student_data = student_data.dropna()

# Remove the invalid 99-hour study value
student_data = student_data[student_data["study_hours"] != 99]

# Split students into two groups based on attendance
high_attendance_scores = student_data[
    student_data["attendance_pct"] >= 80
]["final_score"]

low_attendance_scores = student_data[
    student_data["attendance_pct"] < 80
]["final_score"]

# Calculate the mean score of each group
high_attendance_mean = high_attendance_scores.mean()
low_attendance_mean = low_attendance_scores.mean()

# Calculate the difference between the two means
mean_score_difference = (
    high_attendance_mean - low_attendance_mean
)

# Perform independent-samples t-test
t_statistic, p_value = stats.ttest_ind(
    high_attendance_scores,
    low_attendance_scores
)

# Display results
print("HIGH ATTENDANCE GROUP (>= 80%)")
print("Number of students:", len(high_attendance_scores))
print("Mean final score:", high_attendance_mean)

print("\nLOW ATTENDANCE GROUP (< 80%)")
print("Number of students:", len(low_attendance_scores))
print("Mean final score:", low_attendance_mean)

print("\nDIFFERENCE BETWEEN MEANS")
print("Mean score difference:", mean_score_difference)

print("\nINDEPENDENT-SAMPLES T-TEST")
print("t-statistic:", t_statistic)
print("p-value:", p_value)

# Decision at 0.05 significance level
if p_value < 0.05:
    print("\nDecision: The difference is statistically significant.")
else:
    print("\nDecision: The difference is not statistically significant.")

# Plain-English conclusion
print("\nCONCLUSION")
if p_value < 0.05:
    print(
        f"Students with attendance >= 80% had an average final score "
        f"that was {abs(mean_score_difference):.2f} points different "
        f"from students with attendance < 80%. "
        f"The p-value is below 0.05, so the difference is statistically significant."
    )
else:
    print(
        f"Students with attendance >= 80% and below 80% had a mean score "
        f"difference of {abs(mean_score_difference):.2f} points. "
        f"The p-value is above 0.05, so the difference is not statistically significant."
    )
```

1. What exactly does the null hypothesis say, in your own words?

   The null hypothesis says that there is **no real difference or effect** between the groups.

2. If p = 0.03, what does that number literally mean?

   It means there is a **3% chance of getting results this extreme if the null hypothesis is true**.

3. Your difference is significant — but is it large enough to act on? How do you decide?

   Check the **effect size and practical importance**, not just the p-value. A small p-value does not always mean the difference is large or useful.


## TASK 04__A chi-square test: are two categories related?

```python
import numpy as np
import pandas as pd
from scipy import stats

# Load the CSV file
student_data = pd.read_csv("student_performance (3).csv")

# Remove missing values
student_data = student_data.dropna()

# Remove the invalid 99-hour study value
student_data = student_data[student_data["study_hours"] != 99]

# Create categorical variables
high_attendance = (student_data["attendance_pct"] >= 80).astype(int)
passed_status = student_data["passed"].astype(int)

# Build 2x2 contingency table
contingency_table = np.zeros((2, 2), dtype=int)

for attendance_status, pass_status in zip(high_attendance, passed_status):
    contingency_table[attendance_status, pass_status] += 1

# Chi-square test
chi_square, p_value, degrees_of_freedom, expected_counts = stats.chi2_contingency(
    contingency_table
)

# Display results
print("Contingency Table:")
print(contingency_table)

print("Chi-square statistic:", chi_square)
print("p-value:", p_value)

# Decision at 0.05 significance level
if p_value < 0.05:
    print("Reject the null hypothesis: attendance and passing are related.")
else:
    print("Fail to reject the null hypothesis: attendance and passing appear independent.")
```

1. Why not a t-test?

   A t-test compares **means of numerical data**. Here, both variables are **categorical** (attendance: high/low, passing: yes/no), so Chi-Square is appropriate.

2. What does expected output represent?

   It shows the **counts we would expect if the two variables were independent** (no relationship).

3. Connection to independence:

   Chi-Square checks whether the **observed counts are significantly different from the expected counts under independence**.

   * Small difference → likely independent
   * Large difference → possible association


## Part C — don't get fooled

## TASK 05__See the traps for yourself

```python
import numpy as np
import pandas as pd
from scipy import stats

# Load student performance data
student_data = pd.read_csv("student_performance (3).csv")

# Remove missing values
student_data = student_data.dropna()

# Remove invalid study-hours value
student_data = student_data[student_data["study_hours"] != 99]

# Get mean and standard deviation from the CSV
study_hours_mean = student_data["study_hours"].mean()
study_hours_std = student_data["study_hours"].std()



# 1. TYPE I ERROR

number_of_tests = 1000
significant_results = 0

rng = np.random.default_rng(42)

for test_number in range(number_of_tests):

    # Both groups come from the SAME distribution
    group_one = rng.normal(
        study_hours_mean,
        study_hours_std,
        30
    )

    group_two = rng.normal(
        study_hours_mean,
        study_hours_std,
        30
    )

    # Independent t-test
    t_statistic, p_value = stats.ttest_ind(group_one, group_two)

    # Count significant results
    if p_value < 0.05:
        significant_results += 1

type_one_error_rate = significant_results / number_of_tests

print("TYPE I ERROR")
print("Number of significant tests:", significant_results)
print("Type I error rate:", type_one_error_rate)
print("Percentage:", type_one_error_rate * 100)



# 2. SIGNIFICANCE VS PRACTICAL IMPORTANCE

large_sample_size = 5000
tiny_difference = 0.2

# Create two large groups using the CSV distribution
group_one_large = rng.normal(
    study_hours_mean,
    study_hours_std,
    large_sample_size
)

group_two_large = rng.normal(
    study_hours_mean + tiny_difference,
    study_hours_std,
    large_sample_size
)

# Independent t-test
t_statistic, p_value = stats.ttest_ind(
    group_one_large,
    group_two_large
)

print("\nSIGNIFICANCE VS PRACTICAL IMPORTANCE")
print("Group 1 mean:", group_one_large.mean())
print("Group 2 mean:", group_two_large.mean())
print("Mean difference:", group_two_large.mean() - group_one_large.mean())
print("t-statistic:", t_statistic)
print("p-value:", p_value)

if p_value < 0.05:
    print("Result: Statistically significant")
else:
    print("Result: Not statistically significant")
```

1. 20 independent tests:

   At a 5% significance level, you would expect about **1 significant result** by chance.
   `20 × 0.05 = 1`

2. Why does a huge sample make tiny effects significant?

   A large sample reduces **random error**, making even very small differences easier to detect statistically.

3. When is a significant effect worth acting on?

   Check its **effect size, practical importance, cost, risk, and real-world impact**—not just the p-value.
