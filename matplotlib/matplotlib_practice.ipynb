## TASK 01__Line and scatte

```python

# QUESTION 1: Line Plot - sin(x) and cos(x)

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

fig, ax = plt.subplots(figsize=(8, 5))

x = np.linspace(0,4 * np.pi,200)

y_sin = np.sin(x)
y_cos = np.cos(x)

ax.plot(x, y_sin, label="sin(x)")
ax.plot(x, y_cos, label="cos(x)")

ax.set_title("Sine and Cosine Functions")
ax.set_xlabel("Angle (radians)")
ax.set_ylabel("Function Value")

ax.legend()

plt.show()



# 2. SCATTER PLOT: y ≈ 2x + noise

np.random.seed(42)

num_points = 100

x_values = np.random.uniform(0,10,num_points)

noise = np.random.uniform(0,2,num_points)

y_values = 2* x_values + noise

fig,ax = plt.subplots(figsize=(8,5))

ax.scatter(x_values,y_values, label="Data points")

ax.set_title("Scatter Plot: y = 2x + Noise")
ax.set_xlabel("X Value (units)")
ax.set_ylabel("Y value (units)")

ax.legend()

plt.show()


'''# 3. RULE:
# ============================================================

# Use a line plot when data has a continuous/order relationship
# Example in ML: training loss over epochs.
#
# Use a scatter plot when you want to see the relationship
# between two individual variables.
# Example in ML: study hours vs exam marks.'''



# 4. ENHANCED SCATTER PLOT

np.random.seed(42)

x = np.random.uniform(0,10,100)
noise = np.random.normal(0,2,100)
y = 2 * x + noise

# Third variable -> colour
colour = np.random.uniform(20,200,100)

#Fourth variable -> size
size = np.random.uniform(20,200,100)

fig,ax = plt.subplots(figsize=(8,5))

scatter = ax.scatter(
    x,
    y,
    c=colour,
    s=size,
    alpha=0.7
)

ax.set_title("Enhanced Scatter Plot")
ax.set_xlabel("X value (units)")
ax.set_ylabel("Y value (units)")


fig.colorbar(scatter,ax=ax,label="Third Variable")

plt.show()






```

1. **Why must x and y arrays be the same length in `ax.plot`?**
   Because each **x value needs one corresponding y value** to form a point `(x, y)`.

2. **When would connecting scatter points with lines be misleading?**
   When the points are **independent observations** and have no meaningful order, because lines can falsely suggest a continuous trend.

3. **What does `alpha` do, and why does it help on a crowded scatter?**
   `alpha` controls **transparency**. Lower alpha makes overlapping points easier to see and shows where many points are concentrated.


## TASK 02__Bar chart versus histogram

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. BAR CHART

regions = ["North","South","West","Central"]
sales_amounts = [340, 280, 190, 410]

fig, ax = plt.subplots(figsize=(8,5))

ax.bar(regions, sales_amounts)

ax.set_title("Sales by Region")
ax.set_xlabel("Region")
ax.set_ylabel("Sales (units)")

plt.show()


# 2. HISTOGRAM - 1000 NORMAL DISTRIBUTION VALUES

np.random.seed(42)

customer_ages = np.random.normal(
    loc=35,
    scale=8,
    size=1000
)

fig,ax = plt.subplots(figsize=(8,5))

ax.hist(customer_ages, bins=30, edgecolor="white")

ax.set_title("Distributed of Customer Ages")
ax.set_xlabel("AGE(years)")
ax.set_ylabel("Number of Customer")

#Mean Line
mean_age = customer_ages.mean()

ax.axvline(
    mean_age,
    color="red",
    linestyle="--",
    label=f"Mean = {mean_age:.2f}"

)

ax.legend()

plt.show()


'''# WHY BAR CHART HAS GAPS BUT HISTOGRAM DOES NOT
# ============================================================

# Bar charts represent separate categories, so gaps show
# that the categories are independent.
#
# Histograms represent continuous numerical ranges (bins),
# so the bars touch to show that the ranges are connected.'''



# 3. HISTOGRAM WITH 5, 30 AND 200 BINS

fig,histogram_axes = plt.subplots(1,3,figsize=(15,5))

bin_counts = [5,30,200]

for current_axis, number_of_bins in zip(histogram_axes,bin_counts):
    
    current_axis.hist(
        customer_ages,
        bins=number_of_bins,
        edgecolor="white"
    
    )

    current_axis.axvline(
        mean_age,
        color="red",
        linestyle="--"
    )

    current_axis.set_title(f"Histogram with {number_of_bins}Bins")
    current_axis.set_xlabel("Age(years)")
    current_axis.set_ylabel("Number of Customers")

plt.tight_layout()
plt.show()


# 30 bins tells the story best because it shows the distribution
# clearly without being too coarse or too noisy.





```

#1. What question does a bar chart answer?
Answer: It answers **“How much does each category have?”**

#2. What question does a histogram answer?

Answer: It answers **“How are numerical values distributed?”**

#3. Why do histogram bars touch while bar-chart bars have gaps?

Answer: Histogram bars touch because they represent **continuous numerical ranges**. Bar-chart bars have gaps because they represent **separate categories**.

#4. How does the number of bins change the story the histogram tells?

Answer: **Fewer bins** show less detail, while **more bins** show more detail but can make the histogram noisy.


## TASK 03__Subplots — a grid that tells one story

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x_values = np.linspace(0, 10, 100)
line_values = np.sin(x_values)

np.random.seed(42)
scatter_x = np.random.uniform(0, 10, 50)
scatter_y = 2 * scatter_x + np.random.normal(0, 2, 50)

categories = ["A", "B", "C", "D"]
category_values = [25, 40, 30, 50]

histogram_data = np.random.normal(50, 10, 1000)


# 1. Create a 2 × 2 grid
fig, axes = plt.subplots(2, 2, figsize=(11, 8))


# Top-left: Line plot
axes[0, 0].plot(x_values, line_values)
axes[0, 0].set_title("Line Plot")
axes[0, 0].set_xlabel("X Value")
axes[0, 0].set_ylabel("sin(x)")


# Top-right: Scatter plot
axes[0, 1].scatter(scatter_x, scatter_y)
axes[0, 1].set_title("Scatter Plot")
axes[0, 1].set_xlabel("X Value")
axes[0, 1].set_ylabel("Y Value")


# Bottom-left: Bar chart
axes[1, 0].bar(categories, category_values)
axes[1, 0].set_title("Bar Chart")
axes[1, 0].set_xlabel("Category")
axes[1, 0].set_ylabel("Value")


# Bottom-right: Histogram
axes[1, 1].hist(histogram_data, bins=30, edgecolor="white")
axes[1, 1].set_title("Histogram")
axes[1, 1].set_xlabel("Value")
axes[1, 1].set_ylabel("Frequency")


# Whole figure title
fig.suptitle("Different Types of Matplotlib Plots", fontsize=16)

# Adjust spacing between subplots
plt.tight_layout()

plt.show()


# 2. Create a 1 × 3 grid
fig, axes = plt.subplots(1, 3, figsize=(15, 5))


# In a 1 × 3 grid, axes is a 1-dimensional array,
# so we use one index instead of two indices.

axes[0].plot(x_values, line_values)
axes[0].set_title("Line Plot")

axes[1].scatter(scatter_x, scatter_y)
axes[1].set_title("Scatter Plot")

axes[2].hist(histogram_data, bins=30, edgecolor="white")
axes[2].set_title("Histogram")


plt.tight_layout()
plt.show()
```

#1. Why is `axes` a 2D array for a 2×2 grid but a 1D array for a 1×3?

Answer: A `2×2` grid has **rows and columns**, so we use two indexes: `axes[0,0]`.
A `1×3` grid has only **one row**, so it can be accessed with one index: `axes[0]`.

#2. What exactly does `tight_layout()` adjust?

Answer: It automatically adjusts **spacing between subplots, titles, and labels** so they don't overlap or get cut off.

#3. What is the difference between `ax.set_title()` and `fig.suptitle()`?

Answer:
 `ax.set_title()` → title for **one individual subplot**.
 `fig.suptitle()` → **one main title for the entire figure**.

## TASK 04__Fix the terrible chart

```python
import matplotlib.pyplot as plt

# TERRIBLE CHART
fig, ax = plt.subplots()

ax.bar(
    ["A", "B", "C"],
    [102, 104, 106],
    color=["red", "lime", "cyan"]
)

ax.set_ylim(100, 107)

plt.show()

'''# FLAWS IN THE TERRIBLE CHART:
# 1. No title
# 2. No X-axis label
# 3. No Y-axis label
# 4. No legend
# 5. Clashing/harsh colours
# 6. No gridlines
# 7. Y-axis starts at 100 instead of 0, which exaggerates the differences
# 8. No proper figure size/layout'''



# FIXED / CLEAN CHART
fig, ax = plt.subplots(figsize=(8, 5))

categories = ["A", "B", "C"]
values = [102, 104, 106]

ax.bar(categories, values)

ax.set_title("Values by Category")
ax.set_xlabel("Category")
ax.set_ylabel("Value")

ax.set_ylim(0, 110)

ax.grid(axis="y", linestyle="--", alpha=0.5)

ax.legend(["Values"])

plt.tight_layout()
plt.show()



'''# The most misleading flaw is the Y-axis starting at 100 instead of 0,
# because it makes the small differences between A, B and C look much larger.
#
# Misleading is worse than ugly because an ugly chart is only unpleasant
# to look at, while a misleading chart can cause people to understand
# the data incorrectly.'''
```

#1. Why does starting a bar chart's y-axis at 100 instead of 0 mislead the reader?

Answer:It makes small differences look **much larger than they actually are**, giving a false impression of the data.

#2. Which flaws make the chart ugly, and which make it dishonest?

Answer:
 *Ugly: No title, no labels, clashing colours, no gridlines, poor layout.
*Dishonest/misleading: A **truncated y-axis** that exaggerates differences.

#3. What is the single most important thing a chart must have to be usable?

Answer: **Clarity** — the reader should be able to understand the data correctly and easily.


## TASK 05__Choose the right plot

```python
import matplotlib.pyplot as plt
import numpy as np

# a. MONTHLY REVENUE → LINE PLOT
# Why: A line plot is best for showing changes and trends over time.

months = np.arange(1, 25)
monthly_revenue = [
    420, 450, 470, 490, 510, 530,
    550, 580, 600, 620, 650, 680,
    700, 730, 750, 780, 810, 840,
    860, 890, 920, 950, 980, 1020
]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(months, monthly_revenue, marker="o")

ax.set_title("Monthly Revenue Over Two Years")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue (₹ thousands)")

plt.tight_layout()
plt.show()



# b. HOUSE SIZE vs PRICE → SCATTER PLOT
# Why: A scatter plot shows the relationship between two numerical variables.

np.random.seed(42)

house_size = np.random.uniform(800, 3000, 100)
house_price = 50 + house_size * 0.08 + np.random.normal(0, 30, 100)

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(house_size, house_price)

ax.set_title("House Size vs House Price")
ax.set_xlabel("House Size (sq ft)")
ax.set_ylabel("Price (₹ lakhs)")

plt.tight_layout()
plt.show()



# c. EXAM SCORES → HISTOGRAM
# Why: A histogram shows how numerical values are distributed and spread

exam_scores = np.random.normal(70, 12, 100)

fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(exam_scores, bins=15, edgecolor="white")

ax.set_title("Distribution of Exam Scores")
ax.set_xlabel("Exam Score")
ax.set_ylabel("Number of Students")

plt.tight_layout()
plt.show()



# d. PRODUCT SALES → BAR CHART
# Why: A bar chart makes it easy to compare sales between categories

products = ["Laptop", "Phone", "Tablet", "Watch", "Headphones"]
product_sales = [120, 180, 95, 150, 210]

fig, ax = plt.subplots(figsize=(8, 5))

ax.bar(products, product_sales)

ax.set_title("Sales by Product")
ax.set_xlabel("Product")
ax.set_ylabel("Units Sold")

plt.tight_layout()
plt.show()



# e. SALARY DISTRIBUTIONS → BOX PLOT
#Why: A box plot compares the distribution, median, and spread of multiple groups.

engineering_salaries = np.random.normal(80, 12, 100)
marketing_salaries = np.random.normal(60, 10, 100)
hr_salaries = np.random.normal(55, 8, 100)
finance_salaries = np.random.normal(70, 11, 100)

salary_data = [
    engineering_salaries,
    marketing_salaries,
    hr_salaries,
    finance_salaries
]

fig, ax = plt.subplots(figsize=(8, 5))

ax.boxplot(salary_data)

ax.set_title("Salary Distribution by Department")
ax.set_xlabel("Department")
ax.set_ylabel("Salary (₹ lakhs)")

ax.set_xticklabels(["Engineering", "Marketing", "HR", "Finance"])

plt.tight_layout()
plt.show()


# f. SUBSCRIPTION PROPORTIONS → PIE CHART
#Why: A pie chart shows how each category makes up part of the whole.

subscription_tiers = ["Basic", "Standard", "Premium"]
user_counts = [500, 300, 200]

fig, ax = plt.subplots(figsize=(7, 7))

ax.pie(
    user_counts,
    labels=subscription_tiers,
    autopct="%1.1f%%"
)

ax.set_title("User Subscription Distribution")

plt.show()


# --------------------------------------------------
# HESITATION NOTES
# --------------------------------------------------

# d. Bar chart vs pie chart:
# Bar chart is better because we need to compare which product sells the most.
#
# e. Histogram vs box plot:
# Box plot is better because it allows comparison of multiple distributions
# in a compact way.
#
# f. Pie chart vs bar chart:
# Pie chart is suitable because the question asks for proportions of one whole.
```

#e. Why might a box plot beat four separate histograms?

Answer: A box plot compares the median, spread, and outliers of all
#four departments easily, while four histograms can be harder to compare.


#f. Why might a bar chart be clearer than a pie chart?

Answer: A bar chart makes it easier to compare the exact sizes of
#categories, especially when the proportions are close.


#Which of these six is most likely to be drawn wrong by a beginner?

Answer: (e) is most likely to be drawn wrong.
#Wrong choice: Using four separate histograms instead of a box plot,
#which makes comparing the distributions more difficult.

## Part C — the polished figure

## TASK 06__One report-quality figure

```python
import numpy as np
import matplotlib.pyplot as plt


# 1. Generate the dataset

np.random.seed(42)

number_of_students = 200

study_hours = np.random.uniform(1, 10, number_of_students)
attendance_percentage = np.random.uniform(50, 100, number_of_students)

# Exam score depends mainly on study hours and attendance
exam_scores = (
    30
    + study_hours * 5
    + attendance_percentage * 0.3
    + np.random.normal(0, 5, number_of_students)
)

# Keep scores between 0 and 100
exam_scores = np.clip(exam_scores, 0, 100)

# Pass if exam score >= 50
passed = exam_scores >= 50



# 2. Create 2x2 figure

fig, axes = plt.subplots(2, 2, figsize=(12, 9))


# Panel 1: Scatter plot

axes[0, 0].scatter(
    study_hours[passed],
    exam_scores[passed],
    color="green",
    label="Passed",
    alpha=0.7
)

axes[0, 0].scatter(
    study_hours[~passed],
    exam_scores[~passed],
    color="red",
    label="Failed",
    alpha=0.7
)

axes[0, 0].set_title("Study Hours vs Exam Score")
axes[0, 0].set_xlabel("Study Hours per Day")
axes[0, 0].set_ylabel("Exam Score")
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)



# Panel 2: Histogram

axes[0, 1].hist(
    exam_scores[passed],
    bins=15,
    color="green",
    alpha=0.7,
    label="Passed"
)

axes[0, 1].hist(
    exam_scores[~passed],
    bins=15,
    color="red",
    alpha=0.7,
    label="Failed"
)

axes[0, 1].set_title("Distribution of Exam Scores")
axes[0, 1].set_xlabel("Exam Score")
axes[0, 1].set_ylabel("Number of Students")
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)



# Panel 3: Average Exam Score

average_exam_scores = [
    exam_scores[passed].mean(),
    exam_scores[~passed].mean()
]

axes[1, 0].bar(
    ["Passed", "Failed"],
    average_exam_scores,
    color=["green", "red"]
)

axes[1, 0].set_title("Average Exam Score by Result")
axes[1, 0].set_xlabel("Result")
axes[1, 0].set_ylabel("Average Exam Score")
axes[1, 0].grid(axis="y", alpha=0.3)



# Panel 4: Average Attendance

average_attendance = [
    attendance_percentage[passed].mean(),
    attendance_percentage[~passed].mean()
]

axes[1, 1].bar(
    ["Passed", "Failed"],
    average_attendance,
    color=["green", "red"]
)

axes[1, 1].set_title("Average Attendance by Result")
axes[1, 1].set_xlabel("Result")
axes[1, 1].set_ylabel("Average Attendance (%)")
axes[1, 1].grid(axis="y", alpha=0.3)



# Overall figure title

fig.suptitle(
    "How Do Study Hours and Attendance Relate to Student Exam Performance?",
    fontsize=16
)

plt.tight_layout(rect=[0, 0, 1, 0.96])



# Save the figure

fig.savefig(
    "student_performance_report.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```


1. Does your figure answer one clear question?
   
   Yes. The figure answers one clear question: how study hours and attendance relate to students' exam performance and pass status. The four panels support this question instead of showing unrelated charts.

2. If you removed all the titles and labels, would anyone understand it?

   No. The titles, axis labels, legends, and consistent colours are necessary to understand what each panel represents and what green and red mean.

3. What is the single takeaway?

   The main takeaway is that students with higher study hours generally tend to have higher exam scores, while attendance also provides useful context for understanding pass status.
