# TASK 02 Load, clean, and run full EDA

```python

# Load, Clean, and Run Full EDA

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load Dataset
df = pd.read_csv("ames_housing.csv")

print("FIRST 5 ROWS")
display(df.head())

# 2. First-30-Minutes Checklist
print("\nDataset Shape:", df.shape)
print("\nColumn Names:", df.columns.tolist())
print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nNeighborhood Counts:")
print(df["Neighborhood"].value_counts())

# 3. Clean Data
# Remove exact duplicate rows
df = df.drop_duplicates().copy()

# Fix data types
df["Neighborhood"] = df["Neighborhood"].astype("category")

numeric_columns = [
    "Id", "OverallQual", "GrLivArea", "YearBuilt",
    "Bedrooms", "FullBath", "GarageCars", "LotArea",
    "SalePrice", "LotFrontage"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Fill missing LotFrontage using neighborhood median,
# then use the overall median
df["LotFrontage"] = df.groupby(
    "Neighborhood", observed=True
)["LotFrontage"].transform(
    lambda x: x.fillna(x.median())
)

df["LotFrontage"] = df["LotFrontage"].fillna(
    df["LotFrontage"].median()
)

# Report missing values in other columns
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# Remove rows missing essential analysis values
essential_columns = [
    "SalePrice", "GrLivArea", "OverallQual",
    "YearBuilt", "Neighborhood"
]

df = df.dropna(subset=essential_columns).copy()

print("\nShape After Cleaning:", df.shape)

# 4. Univariate Analysis
# Numerical distributions
numerical_columns = [
    "SalePrice", "GrLivArea", "LotArea",
    "OverallQual", "YearBuilt", "Bedrooms",
    "FullBath", "GarageCars", "LotFrontage"
]

df[numerical_columns].hist(
    figsize=(15, 12),
    bins=25,
    edgecolor="black"
)

plt.suptitle("Univariate Analysis of Housing Features")
plt.tight_layout()
plt.show()

# SalePrice distribution
plt.figure(figsize=(8, 5))
sns.histplot(df["SalePrice"], kde=True, bins=30)
plt.title("Distribution of Sale Price")
plt.tight_layout()
plt.show()

# Neighborhood counts
plt.figure(figsize=(12, 6))
sns.countplot(
    data=df,
    y="Neighborhood",
    order=df["Neighborhood"].value_counts().index
)
plt.title("Houses by Neighborhood")
plt.tight_layout()
plt.show()



# 5. Bivariate Analysis
# Living area vs sale price
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="GrLivArea", y="SalePrice")
plt.title("Living Area vs Sale Price")
plt.tight_layout()
plt.show()

# Overall quality vs sale price
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="OverallQual", y="SalePrice")
plt.title("Overall Quality vs Sale Price")
plt.tight_layout()
plt.show()

# Neighborhood vs sale price
neighborhood_order = (
    df.groupby("Neighborhood", observed=True)["SalePrice"]
    .median()
    .sort_values()
    .index
)

plt.figure(figsize=(12, 7))
sns.boxplot(
    data=df, x="SalePrice", y="Neighborhood",
    order=neighborhood_order
)
plt.title("Neighborhood vs Sale Price")
plt.tight_layout()
plt.show()

# Year built vs sale price
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="YearBuilt", y="SalePrice")
plt.title("Year Built vs Sale Price")
plt.tight_layout()
plt.show()

# Bedrooms, bathrooms, garage capacity vs sale price
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

sns.boxplot(data=df, x="Bedrooms", y="SalePrice", ax=axes[0])
axes[0].set_title("Bedrooms vs Sale Price")

sns.boxplot(data=df, x="FullBath", y="SalePrice", ax=axes[1])
axes[1].set_title("Bathrooms vs Sale Price")

sns.boxplot(data=df, x="GarageCars", y="SalePrice", ax=axes[2])
axes[2].set_title("Garage Capacity vs Sale Price")

plt.tight_layout()
plt.show()

# 6. Multivariate Analysis: Correlation Heatmap
correlation_matrix = df.select_dtypes(
    include="number"
).drop(columns=["Id"], errors="ignore").corr()

plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

# Living area, sale price, and quality together
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="GrLivArea",
    y="SalePrice",
    hue="OverallQual",
    palette="viridis"
)
plt.title("Living Area vs Sale Price by Quality")
plt.tight_layout()
plt.show()

# 7. Outlier Detection Using IQR
outlier_columns = [
    "SalePrice", "GrLivArea", "LotArea", "LotFrontage"
]

outlier_results = []

for column in outlier_columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    df[f"{column}_outlier"] = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    )

    outlier_results.append({
        "Column": column,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": int(df[f"{column}_outlier"].sum())
    })

print("\nOutlier Summary:")
display(pd.DataFrame(outlier_results))

# Outlier boxplots
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, column in zip(axes.flatten(), outlier_columns):
    sns.boxplot(x=df[column], ax=ax)
    ax.set_title(f"Outliers in {column}")

plt.tight_layout()
plt.show()

# 8. Final Summary
print("\nFINAL DATASET SHAPE:", df.shape)
print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nSale Price Summary:")
print(df["SalePrice"].describe())

print("\nCleaning Decisions:")
print("- Removed exact duplicate rows.")
print("- Filled missing LotFrontage with neighborhood and overall medians.")
print("- Converted columns to appropriate data types.")
print("- Removed rows missing essential analysis values.")
print("- Detected and flagged potential outliers using IQR.")
print("- Kept outliers for investigation instead of deleting them automatically.")

# 9. Save Cleaned Dataset
df.to_csv("ames_housing_cleaned.csv", index=False)

print("\nCleaned dataset saved as ames_housing_cleaned.csv")
```

## TASK 03 Test two hypotheses

```python

import pandas as pd
import numpy as np
from scipy import stats

# Load dataset
df = pd.read_csv("ames_housing.csv")

# Keep required columns and remove missing values
df = df.dropna(
    subset=["SalePrice", "OverallQual", "Bedrooms"]
).copy()



# HYPOTHESIS 1: T-TEST

# Compare average prices of high-quality and other houses
high_quality = df.loc[df["OverallQual"] >= 7, "SalePrice"]
other_quality = df.loc[df["OverallQual"] < 7, "SalePrice"]

# Null hypothesis: Both groups have equal average sale prices.
# Alternative hypothesis: Their average sale prices differ.

t_stat, p_value = stats.ttest_ind(
    high_quality, other_quality, equal_var=False
)

mean_high = high_quality.mean()
mean_other = other_quality.mean()
mean_difference = mean_high - mean_other

# Cohen's d effect size
pooled_sd = np.sqrt(
    (high_quality.var(ddof=1) + other_quality.var(ddof=1)) / 2
)
cohens_d = mean_difference / pooled_sd if pooled_sd else np.nan

print("HYPOTHESIS 1: T-TEST")
print("H0: Both groups have equal average sale prices.")
print("High-quality houses (quality >= 7):", round(mean_high, 2))
print("Other houses (quality < 7):", round(mean_other, 2))
print("Mean difference:", round(mean_difference, 2))
print("T-statistic:", round(t_stat, 4))
print("P-value:", p_value)
print("Cohen's d:", round(cohens_d, 4))

if p_value < 0.05:
    print("Decision: Reject H0. The average prices differ significantly.")
else:
    print("Decision: Fail to reject H0. There is insufficient evidence of a difference.")

if abs(cohens_d) < 0.2:
    print("Effect size: Small or negligible.")
elif abs(cohens_d) < 0.5:
    print("Effect size: Small.")
elif abs(cohens_d) < 0.8:
    print("Effect size: Medium.")
else:
    print("Effect size: Large.")



# HYPOTHESIS 2: CHI-SQUARE TEST

# Group bedrooms into two categories
df["BedroomGroup"] = np.where(
    df["Bedrooms"] <= 2, "2 or fewer bedrooms", "3 or more bedrooms"
)

# Group overall quality into two categories
df["QualityGroup"] = np.where(
    df["OverallQual"] >= 7, "High quality", "Other quality"
)

# Null hypothesis: Bedroom group and quality group are independent.
# Alternative hypothesis: They are associated.

table = pd.crosstab(df["BedroomGroup"], df["QualityGroup"])

chi2, p_chi, dof, expected = stats.chi2_contingency(table)

# Cramer's V effect size
n = table.to_numpy().sum()
min_dimension = min(table.shape[0] - 1, table.shape[1] - 1)
cramers_v = np.sqrt(chi2 / (n * min_dimension))

print("\nHYPOTHESIS 2: CHI-SQUARE TEST")
print("H0: Bedroom group and quality group are independent.")
print("\nObserved counts:")
print(table)
print("\nChi-square statistic:", round(chi2, 4))
print("Degrees of freedom:", dof)
print("P-value:", p_chi)
print("Cramer's V:", round(cramers_v, 4))
print("Any expected count below 5:", (expected < 5).any())

if p_chi < 0.05:
    print("Decision: Reject H0. Bedroom group and quality group are associated.")
else:
    print("Decision: Fail to reject H0. There is insufficient evidence of an association.")

if cramers_v < 0.1:
    print("Effect size: Small or negligible.")
elif cramers_v < 0.3:
    print("Effect size: Small to moderate.")
elif cramers_v < 0.5:
    print("Effect size: Moderate.")
else:
    print("Effect size: Large.")
```

1\. Does your test match your data type? 

Yes. The t-test compares average sale prices between two groups, while the chi-square test checks the association between categorical variables.

2\. If a result is significant but tiny, how do you report it honestly?

 Report that the result is statistically significant but has a small practical effect, using an effect size and the actual difference.

3\. What would failing to reject the null actually tell you? 

There is insufficient evidence to reject the null hypothesis. It does not prove that the null hypothesis is true.

```python

# TASK: Ames Housing - Complete EDA Visualizations

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set a consistent visual theme and colour palette
sns.set_theme(style="whitegrid", palette="deep")

# 1. Load the dataset
df = pd.read_csv("ames_housing.csv")

# Basic cleaning for visualization
df = df.drop_duplicates().copy()

numeric_columns = [
    "OverallQual", "GrLivArea", "YearBuilt",
    "Bedrooms", "FullBath", "GarageCars",
    "LotArea", "SalePrice", "LotFrontage"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Fill missing LotFrontage using neighborhood median,
# then overall median
df["LotFrontage"] = df.groupby(
    "Neighborhood", observed=True
)["LotFrontage"].transform(
    lambda x: x.fillna(x.median())
)
df["LotFrontage"] = df["LotFrontage"].fillna(
    df["LotFrontage"].median()
)

# Remove rows missing values required for the plots
df = df.dropna(
    subset=["SalePrice", "GrLivArea", "OverallQual", "Neighborhood"]
).copy()


# 1. UNIVARIATE: SALE PRICE DISTRIBUTION

plt.figure(figsize=(8, 5))
sns.histplot(
    data=df, x="SalePrice", bins=30,
    kde=True, color="steelblue"
)
plt.title("1. Distribution of House Sale Prices")
plt.xlabel("Sale Price (dataset currency units)")
plt.ylabel("Number of Houses")
plt.tight_layout()
plt.show()


# 2. UNIVARIATE: LIVING AREA DISTRIBUTION

plt.figure(figsize=(8, 5))
sns.histplot(
    data=df, x="GrLivArea", bins=30,
    kde=True, color="darkorange"
)
plt.title("2. Distribution of Above-Ground Living Area")
plt.xlabel("Living Area (square feet, if using standard Ames data)")
plt.ylabel("Number of Houses")
plt.tight_layout()
plt.show()


# 3. UNIVARIATE: HOUSE QUALITY COUNTS

plt.figure(figsize=(8, 5))
sns.countplot(
    data=df, x="OverallQual",
    color="seagreen"
)
plt.title("3. Number of Houses by Overall Quality")
plt.xlabel("Overall Quality Score")
plt.ylabel("Number of Houses")
plt.tight_layout()
plt.show()


# 4. BIVARIATE: LIVING AREA VS PRICE

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df, x="GrLivArea", y="SalePrice",
    color="royalblue", alpha=0.65
)
plt.title("4. Relationship Between Living Area and Sale Price")
plt.xlabel("Living Area (square feet, if using standard Ames data)")
plt.ylabel("Sale Price (dataset currency units)")
plt.tight_layout()
plt.show()


# 5. BIVARIATE: QUALITY VS PRICE
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df, x="OverallQual", y="SalePrice",
    color="mediumpurple"
)
plt.title("5. Sale Price Across House Quality Scores")
plt.xlabel("Overall Quality Score")
plt.ylabel("Sale Price (dataset currency units)")
plt.tight_layout()
plt.show()


# 6. BIVARIATE: NEIGHBORHOOD VS PRICE
neighborhood_order = (
    df.groupby("Neighborhood", observed=True)["SalePrice"]
    .median()
    .sort_values()
    .index
)

plt.figure(figsize=(11, 7))
sns.boxplot(
    data=df, x="SalePrice", y="Neighborhood",
    order=neighborhood_order,
    color="cornflowerblue"
)
plt.title("6. Sale Price Distribution by Neighborhood")
plt.xlabel("Sale Price (dataset currency units)")
plt.ylabel("Neighborhood")
plt.tight_layout()
plt.show()


# 7. BIVARIATE: YEAR BUILT VS PRICE

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df, x="YearBuilt", y="SalePrice",
    color="teal", alpha=0.65
)
plt.title("7. Construction Year vs Sale Price")
plt.xlabel("Year Built")
plt.ylabel("Sale Price (dataset currency units)")
plt.tight_layout()
plt.show()


# 8. MULTIVARIATE: CORRELATION HEATMAP

corr_columns = [
    "OverallQual", "GrLivArea", "YearBuilt",
    "Bedrooms", "FullBath", "GarageCars",
    "LotArea", "SalePrice", "LotFrontage"
]

correlation_matrix = df[corr_columns].corr()

plt.figure(figsize=(11, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)
plt.title("8. Correlation Between Housing Features")
plt.xlabel("Housing Features")
plt.ylabel("Housing Features")
plt.tight_layout()
plt.show()


# 9. MULTIVARIATE: AREA, PRICE AND QUALITY
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df, x="GrLivArea", y="SalePrice",
    hue="OverallQual", palette="viridis",
    alpha=0.75
)
plt.title("9. Living Area vs Sale Price by House Quality")
plt.xlabel("Living Area (square feet, if using standard Ames data)")
plt.ylabel("Sale Price (dataset currency units)")
plt.legend(title="Overall Quality")
plt.tight_layout()
plt.show()


# 10. OUTLIERS: SALE PRICE BOXPLOT
plt.figure(figsize=(8, 4))
sns.boxplot(
    data=df, x="SalePrice",
    color="salmon"
)
plt.title("10. Outlier Detection in House Sale Prices")
plt.xlabel("Sale Price (dataset currency units)")
plt.ylabel("Distribution")
plt.tight_layout()
plt.show()


# FINAL SUMMARY
print("Dataset shape:", df.shape)
print("\nAverage sale price:", round(df["SalePrice"].mean(), 2))
print("Median sale price:", round(df["SalePrice"].median(), 2))
print("Average living area:", round(df["GrLivArea"].mean(), 2))

print("\nTop 5 neighborhoods by median sale price:")
print(
    df.groupby("Neighborhood", observed=True)["SalePrice"]
    .median()
    .sort_values(ascending=False)
    .head(5)
)
```
