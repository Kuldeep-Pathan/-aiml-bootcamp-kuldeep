## TASK 01__Statistics along axes, on multi-dimensional data

````python
import numpy as np

# 1. Create the 3D sales dataset

rng = np.random.default_rng(42)

#Shape:
#4 shops x 30 days x 3 product categories
sales = rng.integers(
    50,
    500,
    size=(4, 30, 3)
)

print("Original sales shape:",sales.shape)
#Expected: (4, 30, 3)


# 2. Total sales per shop

'''#Collapse:
# axis=1 -> days
# axis=2 -> product categories'''

per_shop_total_sales = sales.sum(axis=(1,2))

print("\nTotal sales per shop:")
print(per_shop_total_sales)

print("Shape:",per_shop_total_sales.shape)
#Expected: (4,)

'''#Each value represents one shop's total sales
# across all 30 days and all 3 categories.'''



# 3. Average daily sales per category
#    across all shops and days

''' 
Collapse:
axis=0 -> shops
axis=1 -> days

We keep axis=2 because we want
the average for each product category.
'''

average_sales_per_category = sales.mean(axis=(0, 1))

print("\nAverage daily sales per category:")
print(average_sales_per_category)

print("Shape:", average_sales_per_category.shape)


# Expected: (3,)

''' 
Each value represents the average sales
of one product category across all shops
and all 30 days.
'''


# 4. Total sales for each day of each shop

'''
Collapse:
axis=2 -> product categories

We keep:
axis=0 -> shops
axis=1 -> days
'''

daily_total_sales = sales.sum(axis=2)

print("\nDaily total sales:")
print(daily_total_sales)

print("Shape:", daily_total_sales.shape)
# Expected: (4, 30)

'''
Each row represents one shop.
Each column represents one day.
Each value is the total sales of all
3 product categories for that shop on that day.
'''


# 5. Best day for each shop

'''
We already calculated daily_total_sales.

Now:
axis=1 -> days

argmax() finds the position of the
highest sales value.
'''

best_day_per_shop = daily_total_sales.argmax(axis=1)

print("\nBest day for each shop:")
print(best_day_per_shop)

print("Shape:", best_day_per_shop.shape)
# Expected: (4,)

'''
There are 4 values because there are 4 shops.

IMPORTANT:
argmax() uses zero-based indexing.

For example:
0 = Day 1
1 = Day 2
2 = Day 3
...
29 = Day 30
'''


# 6. Standard deviation of daily sales

'''
We want to check how consistent each shop is.

axis=1 -> calculate standard deviation
across the 30 days.

Lower standard deviation means
more consistent sales.
'''

sales_std_per_shop = daily_total_sales.std(axis=1)

print("\nStandard deviation per shop:")
print(sales_std_per_shop)

print("Shape:", sales_std_per_shop.shape)
# Expected: (4,)


# 7. Find the most consistent shop

'''
argmin() finds the position of the
smallest standard deviation.
'''

most_consistent_shop = sales_std_per_shop.argmin()

print("\nMost consistent shop:", most_consistent_shop)

'''
The result is between 0 and 3 because
we have 4 shops.

0 = Shop 1
1 = Shop 2
2 = Shop 3
3 = Shop 4
'''


# 8. Calculate shop mean using keepdims=True

'''
Collapse:
axis=1 -> days
axis=2 -> product categories

keepdims=True keeps the collapsed
dimensions as size 1.
'''

shop_mean_sales = sales.mean(
    axis=(1, 2),
    keepdims=True
)

print("\nShop mean sales:")
print(shop_mean_sales)

print("Shape:", shop_mean_sales.shape)
# Expected: (4, 1, 1)


# 9. Mean-centre the original sales data

'''
Subtract each shop's mean from
its original sales values.

Because shop_mean_sales has shape
(4, 1, 1), NumPy broadcasts it
across 30 days and 3 categories.
'''

mean_centered_sales = sales - shop_mean_sales

print("\nMean-centred sales:")
print(mean_centered_sales)

print("Shape:", mean_centered_sales.shape)
# Expected: (4, 30, 3)


````

1. For shape `(4, 30, 3)`, what does `axis=1` collapse?

   `axis=1` collapses **30 days**, leaving shops and categories.
   Result shape:* (4, 3)`

2. When do you need `keepdims=True`?

   When you want to keep the collapsed axis as size `1`, mainly for broadcasting. Without it, the result may have an incompatible shape for operations with the original array.

3. Can you pass a tuple to `axis`?

   Yes. `axis=(1,2)` means collapse **both days and categories**.
   For `(4,30,3)`, the result shape is `(4,)`.


## TASK 02__Solve a linear system with Ax = b

````python
import numpy as np
import time


# 1. Write the equation as matrices

coefficient_matrix = np.array([
    [2.,  3., -1.],
    [4., -1.,  2.],
    [-1., 2.,  3.]
])

constant_values = np.array([5., 6., 7.])

print("coefficient matrix A:")
print(coefficient_matrix)

print("\nConstant values b:")
print(constant_values)


# 2. Solve using np.linalg.solve()

solution_vector = np.linalg.solve(
    coefficient_matrix,
    constant_values
)

print("\nSolution [x, y, z]:")
print(solution_vector)


# 3. Verify the solution

calculated_values = coefficient_matrix @ solution_vector

print("\nA @ x:")
print(calculated_values)

# allclose is better than == because floating-point
# calculations can have very small rounding errors.
print(
    "\nDoes A @ x equal b?",
    np.allclose(calculated_values, constant_values)
)


# 4. Solve using inverse matrix

inverse_matrix = np.linalg.inv(coefficient_matrix)

inverse_solution = inverse_matrix @ constant_values

print("\nSolution using inv(A) @ B:")
print(inverse_solution)

# Compare both solutions
print(
    "\n Are bpth solutions the same?",
    np.allclose(solution_vector, inverse_solution)
)



# 5. Time np.linalg.solve()

random_generator = np.random.default_rng(42)

large_coefficient_matrix = random_generator.random((500,500))
large_constant_values = random_generator.random(500)

start_time = time.perf_counter()

start_result = np.linalg.solve(
    large_coefficient_matrix,
    large_constant_values
)

solve_time = time.perf_counter() - start_time

print("\nnp.linalg.solve() time:",solve_time)


# 6. Time inv(A) @ b

start_time = time.perf_counter()

large_inverse_matrix = np.linalg.inv(large_coefficient_matrix)
inverse_result = large_inverse_matrix @ large_constant_values

inverse_time = time.perf_counter() - start_time

print("inv(A) @ b time:", inverse_time)


# Ratio
time_ratio = inverse_time / solve_time

print("Time ratio (inv / solve):", time_ratio)


# 7. Create a singular matrix

singular_matrix = np.array([
    [1., 2., 3.],
    [2., 4., 6.],   # 2 × first row
    [4., 5., 6.]
])

singular_constants = np.array([6., 12., 15.])

print("\nSingular matrix:")
print(singular_matrix)


# Try solving the singular system
try:
    singular_solution = np.linalg.solve(
        singular_matrix,
        singular_constants
    )

    print(singular_solution)

except np.linalg.LinAlgError as error:
    print("\nError:", error)

    # Singular matrix has no unique solution because
    # its rows are linearly dependent.

````

   - `solve()` is **faster** because it avoids calculating the inverse.
   - `solve()` is **more numerically stable/accurate** than explicitly using `inv(A) @ b`.

2. What does a determinant of zero tell you about the system's solutions?

   `det(A) = 0` means the matrix is singular, so the system does not have a unique solution. It may have no solution or infinitely many solutions.

3. In ML, what does a singular matrix usually mean about your features?

   It usually means some features are linearly dependent/redundant—one feature can be expressed from others.

## TASK 03__Cosine similarity from scratch

````python
import numpy as np


# 1. Cosine similarity function


def cosine_similarity(vector_a: np.ndarray,
            vector_b: np.ndarray) -> float:
    """
    Calculate cosine similarity between two vectors.
    """

    dot_product = vector_a @ vector_b

    vector_a_norm = np.linalg.norm(vector_a)
    vector_b_norm = np.linalg.norm(vector_b)

    similarity = dot_product / (vector_a_norm * vector_b_norm)

    return similarity



# 2. Test the three known cases


# Identical vectors → similarity = 1
vector = np.array([1., 2., 3.])

assert np.isclose(
    cosine_similarity(vector, vector),
    1.0
)


# Opposite vectors → similarity = -1
opposite_vector = -vector

assert np.isclose(
    cosine_similarity(vector, opposite_vector),
    -1.0
)


# Perpendicular vectors → similarity = 0
perpendicular_vector_a = np.array([1., 0.])
perpendicular_vector_b = np.array([0., 1.])

assert np.isclose(
    cosine_similarity(
        perpendicular_vector_a,
        perpendicular_vector_b
    ),
    0.0
)

print("All three tests passed!")




# 3. Create 100 vectors

random_generator = np.random.default_rng(42)

vector_matrix = random_generator.random((100, 10))

print("Vector matrix shape:", vector_matrix.shape)
# Expected: (100, 10)



# 4. Similarity of every vector with ONE query vector
query_vector = vector_matrix[0]

# Calculate norm of every row
row_norms = np.linalg.norm(
    vector_matrix,
    axis=1,
    keepdims=True
)

# Normalize every row
normalized_vectors = vector_matrix / row_norms

# Normalize query vector
normalized_query_vector = (
    query_vector / np.linalg.norm(query_vector)
)

# Calculate similarity with every vector
similarity_with_query = (
    normalized_vectors @ normalized_query_vector
)

print("Similarity with query shape:",
        similarity_with_query.shape)
# Expected: (100,)



# 5. Full 100 × 100 similarity matrix

# Every row has length 1
normalized_vector_matrix = (
    vector_matrix / row_norms
)

# Matrix multiplication gives every pair's similarity
full_similarity_matrix = (
    normalized_vector_matrix @ normalized_vector_matrix.T
)

print("Full similarity matrix shape:",
        full_similarity_matrix.shape)
# Expected: (100, 100)



# 6. Why cosine similarity for text embeddings?

# Cosine similarity compares the direction of vectors
# rather than their magnitude. This is useful for text
# embeddings because two texts can have similar meaning
# even when their embedding magnitudes are different.
````

 1. `np.linalg.norm` and `axis`

`np.linalg.norm()` calculates the **magnitude/length** of a vector.
`axis` decides **which direction** to calculate it:

- `axis=0` → columns
- `axis=1` → rows

2. Why `keepdims=True`?

It keeps the result shape as `(100, 1)` instead of `(100,)`, so NumPy can **broadcast it correctly** when normalising each row.

 3. Cosine similarity range

Cosine similarity ranges from **-1 to 1**:

 1 → same direction
  0 → perpendicular
 -1 → opposite direction


## TASK 04__Generate synthetic 2D data with controllable noise

````python
import numpy as np
import matplotlib.pyplot as plt


# 1. Create noisy straight-line data

def make_line(n, slope, intercept, noise, seed=0):
    random_generator = np.random.default_rng(seed)

    x_values = np.linspace(0, 10, n)

    noise_values = random_generator.normal(0, noise, n)

    y_values = slope * x_values + intercept + noise_values

    return x_values, y_values


# 2. Same slope, different noise levels
noise_levels = [0.5, 2.0, 10.0]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for axis, noise_level in zip(axes, noise_levels):

    x_values, y_values = make_line(
        n=100,
        slope=2,
        intercept=3,
        noise=noise_level,
        seed=42
    )

    axis.scatter(x_values, y_values)

    estimated_slope, estimated_intercept = np.polyfit(
        x_values,
        y_values,
        1
    )

    axis.set_title(
        f"Noise = {noise_level}\n"
        f"Estimated slope = {estimated_slope:.2f}"
    )

    axis.set_xlabel("X")
    axis.set_ylabel("Y")

plt.tight_layout()
plt.show()


# 3.
'''Noise = 0.5  → slope recovery very reliable
Noise = 2.0  → reasonably reliable
Noise = 10.0 → much less reliable

So noise = 10.0 makes slope recovery the most unreliable.'''


# 4. Create Cluster

def make_clusters(n, centres, spread, seed=0):
    random_generator = np.random.default_rng(seed)

    cluster_points = []

    for centre_x, centre_y in centres:

        points = random_generator.normal(
            loc=[centre_x, centre_y],
            scale=spread,
            size=(n, 2)
        )

        cluster_points.append(points)

    return cluster_points


# 5. Clearly overlapping clusters

overlapping_clusters = make_clusters(
    n=100,
    centres=[(2, 2), (3, 3)],
    spread=1.5,
    seed=42
)

plt.figure(figsize=(6, 5))

plt.scatter(
    overlapping_clusters[0][:, 0],
    overlapping_clusters[0][:, 1],
    label="Cluster 1"
)

plt.scatter(
    overlapping_clusters[1][:, 0],
    overlapping_clusters[1][:, 1],
    label="Cluster 2"
)

plt.title("Overlapping Clusters")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()


#Clearly separated clusters

separated_clusters = make_clusters(
    n=100,
    centres=[(2, 2), (8, 8)],
    spread=0.8,
    seed=42
)

plt.figure(figsize=(6, 5))

plt.scatter(
    separated_clusters[0][:, 0],
    separated_clusters[0][:, 1],
    label="Cluster 1"
)

plt.scatter(
    separated_clusters[1][:, 0],
    separated_clusters[1][:, 1],
    label="Cluster 2"
)

plt.title("Separated Clusters")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
````

1. `rng.normal(0, noise, n)`

   - `0` → mean
   - `noise` → standard deviation (spread)
   - `n` → number of values

2. Why higher noise makes slope recovery unreliable?

   More noise scatters points away from the true line, making the estimated slope less accurate. It becomes unreliable when noise is large compared with the signal.

3. Why use synthetic data?
   It lets us control the data and know the true answer , making it easier to test and understand algorithms.

## TASK 05__Break things on purpose

````python
import numpy as np


# 1. Floating-point representation error

decimal_sum = 0.1 + 0.2

print("0.1 + 0.2 =", repr(decimal_sum))
print("Is it exactly 0.3?", decimal_sum == 0.3)

'''# Why?
# Binary floating-point cannot represent many decimal
# fractions exactly.
# Detect: use np.isclose()
# Avoid: use np.isclose() or Decimal when exact decimals are needed.'''


# 2. Two more decimal examples

first_decimal_sum = 0.1 + 0.7
second_decimal_sum = 0.2 + 0.4

print("\nMore floating-point examples:")
print("0.1 + 0.7 =", repr(first_decimal_sum))
print("0.2 + 0.4 =", repr(second_decimal_sum))

'''# Detect: compare using np.isclose()
# Avoid: use Decimal for exact decimal arithmetic.'''



# 3. Integer overflow with int8

small_integer_values = np.array(
    [100, 120, 127],
    dtype=np.int8
)

overflow_result = small_integer_values * 2

print("\nint8 overflow:")
print("Original:", small_integer_values)
print("After multiplication:", overflow_result)

'''# int8 range is -128 to 127.
# Values above 127 overflow and can become negative.
# Detect: check dtype and possible value range.
# Avoid: use a larger integer type such as int16 or int64.'''



# 4. float32 precision loss

large_float32_value = np.float32(1e8)

float32_difference = (
    np.float32(large_float32_value + 1)
    - large_float32_value
)

print("\nfloat32 precision loss:")
print("x =", large_float32_value)
print("x + 1 - x =", float32_difference)

'''# At this size, float32 cannot represent every integer.
# Adding 1 may make no difference.
# Detect: test precision at the expected data scale.
# Avoid: use float64 when higher precision is required.'''



# 5. Catastrophic cancellation

large_float32_number = np.float32(1e8)
slightly_larger_float32_number = np.float32(1e8 + 10)

cancellation_result = (
    slightly_larger_float32_number
    - large_float32_number
)

print("\nCatastrophic cancellation:")
print("Expected difference: 10")
print("float32 difference:", cancellation_result)

'''# Subtracting nearly equal large numbers can lose precision.
# Detect: compare results with higher precision.
# Avoid: reformulate the calculation or use float64.'''



# 6. Summing one million small floats

number_of_values = 1_000_000
small_float_value = 0.1

small_float_array = np.full(
    number_of_values,
    small_float_value,
    dtype=np.float32
)

exact_sum = number_of_values * small_float_value

float32_sum = np.sum(
    small_float_array,
    dtype=np.float32
)

float64_sum = np.sum(
    small_float_array,
    dtype=np.float64
)

float32_error = abs(float32_sum - exact_sum)
float64_error = abs(float64_sum - exact_sum)

print("\nLarge summation:")
print("Exact answer:", exact_sum)
print("float32 sum:", float32_sum)
print("float64 sum:", float64_sum)

print("float32 error:", float32_error)
print("float64 error:", float64_error)

print(
    "Closer result:",
    "float32" if float32_error < float64_error else "float64"
)

'''# Detect: compare numerical error with a known/reference value.
# Avoid: use float64 or a numerically stable summation method
# when higher accuracy is required.'''
````

1. Why can't binary floating point represent `0.1` exactly?

   Because `0.1` has a repeating representation in **binary**, so the computer stores a close approximation instead.

2. What is machine epsilon?

   Machine epsilon is the **smallest difference between 1.0 and the next representable floating-point number**.

   ```python
   np.finfo(np.float64).eps
   ```

3. Why does summation order affect the result?

   Floating-point calculations have rounding errors. Adding numbers in different orders can accumulate these errors differently, producing slightly different results.

## TASK 06__Build PCA by hand

````python
import numpy as np
import matplotlib.pyplot as plt


# 1. Generate 300 samples with 4 features

random_generator = np.random.default_rng(42)

number_of_samples = 300

# Base random variables
feature_one = random_generator.normal(0, 1, number_of_samples)

# Feature 2 strongly correlated with Feature 1
feature_two = (
    0.9 * feature_one
    + 0.4 * random_generator.normal(0, 1, number_of_samples)
)

# Feature 3 weakly correlated with Feature 1
feature_three = (
    0.2 * feature_one
    + 0.98 * random_generator.normal(0, 1, number_of_samples)
)

# Feature 4 is independent
feature_four = random_generator.normal(0, 1, number_of_samples)

# Combine all features
data_matrix = np.column_stack([
    feature_one,
    feature_two,
    feature_three,
    feature_four
])

print("Original data shape:", data_matrix.shape)
# Expected: (300, 4)



# 2. Standardise every feature

feature_means = data_matrix.mean(axis=0)
feature_standard_deviations = data_matrix.std(axis=0)

standardized_data = (
    data_matrix - feature_means
) / feature_standard_deviations

print("\nStandardized data shape:", standardized_data.shape)

# Verify mean ≈ 0
print(
    "Mean is approximately 0:",
    np.allclose(
        standardized_data.mean(axis=0),
        0
    )
)

# Verify standard deviation ≈ 1
print(
    "Standard deviation is approximately 1:",
    np.allclose(
        standardized_data.std(axis=0),
        1
    )
)



# 3. Calculate covariance matrix

covariance_matrix = (
    standardized_data.T @ standardized_data
) / (number_of_samples - 1)

print("\nCovariance matrix:")
print(covariance_matrix)

print("Covariance matrix shape:", covariance_matrix.shape)
# Expected: (4, 4)

# A covariance matrix should be symmetric
print(
    "Is covariance matrix symmetric?",
    np.allclose(
        covariance_matrix,
        covariance_matrix.T
    )
)



# 4. Find eigenvalues and eigenvectors

eigenvalues, eigenvectors = np.linalg.eigh(
    covariance_matrix
)

# np.linalg.eigh() returns eigenvalues in ascending order.
# We need descending order.

descending_order = np.argsort(eigenvalues)[::-1]

sorted_eigenvalues = eigenvalues[descending_order]
sorted_eigenvectors = eigenvectors[:, descending_order]

print("\nSorted eigenvalues:")
print(sorted_eigenvalues)

print("\nSorted eigenvectors:")
print(sorted_eigenvectors)



# 5. Calculate explained variance ratio

total_variance = np.sum(sorted_eigenvalues)

explained_variance_ratio = (
    sorted_eigenvalues / total_variance
)

print("\nExplained variance ratio:")
print(explained_variance_ratio)

print("\nExplained variance percentage:")
print(explained_variance_ratio * 100)


# Cumulative explained variance
cumulative_variance_ratio = np.cumsum(
    explained_variance_ratio
)

print("\nCumulative explained variance:")
print(cumulative_variance_ratio)

# Number of components needed for 90% variance
components_for_90_percent = (
    np.argmax(cumulative_variance_ratio >= 0.90) + 1
)

print(
    "\nComponents needed for 90% variance:",
    components_for_90_percent
)



# 6. Project data onto first two principal components

first_two_eigenvectors = sorted_eigenvectors[:, :2]

reduced_data = (
    standardized_data @ first_two_eigenvectors
)

print("\nReduced data shape:", reduced_data.shape)
# Expected: (300, 2)



# 7. Plot the first two principal components

plt.figure(figsize=(8, 6))

plt.scatter(
    reduced_data[:, 0],
    reduced_data[:, 1]
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Data Projected onto First Two Principal Components")

plt.tight_layout()
plt.show()


# 8. Compare with np.cov()

numpy_covariance_matrix = np.cov(
    standardized_data.T
)

print("\nCovariance matrix from np.cov():")
print(numpy_covariance_matrix)

print(
    "\nDo both covariance matrices match?",
    np.allclose(
        covariance_matrix,
        numpy_covariance_matrix
    )
)


# Compare eigenvalues with covariance matrix from np.cov()
numpy_covariance_eigenvalues = np.linalg.eigvalsh(
    numpy_covariance_matrix
)

numpy_covariance_eigenvalues = np.sort(
    numpy_covariance_eigenvalues
)[::-1]

print(
    "\nDo eigenvalues also match?",
    np.allclose(
        sorted_eigenvalues,
        numpy_covariance_eigenvalues
    )
)
````

1. Why standardise before PCA?

   To put all features on the **same scale**. Otherwise, a feature with values in millions would dominate PCA over a feature with small decimal values.

2. Why divide covariance by `n-1` instead of `n`?

   `n-1` gives an **unbiased estimate of population variance** when calculating covariance from a sample.

3. What does the largest eigenvalue/eigenvector mean?

   The eigenvector with the **largest eigenvalue** represents the direction in which the data has the **maximum variance**. That's why it is called the **first principal component**.
