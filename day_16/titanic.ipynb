## Pandas Fundamentals — the Titanic

## Part A — answer 15 questions with Pandas (the deliverable)

## TASK 01 The 15 Titanic questions

```python
import pandas as pd


# load dataset
df = pd.read_csv("titanic (1).csv")


# Quike Inspection
print("Titanic:")
display(df.head())
display(df.info())


# 1. How many passengers are in the dataset?
print("1. Total passengers:", len(df))


# 2. What fraction of passengers survived overall?
print("\n2. overall survial rate:", df["survived"].mean)


# 3. Survival rate for women versus men
print("\n3. Survival rate by sex:")
print(df.groupby("sex")["survived"].mean())


# 4. Survival rate for each passenger class
print("\n4. Survival rate by class:")
print(df.groupby("pclass")["survived"].mean())


# 5. Average age of passengers
print("\n5. Average age:", df["age"].mean())


# 6. Oldest and Youngest Passenger
print("\n6. Oldest passenger age:", df["age"].max())
print(" youngest passenger age:", df["age"].min())


# 7. Average fare paid in each class
print("\n7. Average fare by class:")
print(df.groupby("pclass")["fare"].mean())


# 8. Number of children under 18
children = df[df["age"]<18]
print("\n8. Number of children:", len(children))


# 9. Survival rate of among children
print("\n9. Childrens survival rate:", children ["survived"].mean())


# 10. Number of passengers from each port
print("\n10. Passengers by port:")
print(df["embarked"].value_counts())


# 11. Number of passengers with missing age
print("\n11. Missing age:", df["age"].isna().sum())



# 12. Survival rate for women in 1st class
women_first = df[
    (df["sex"] == "female") &
    (df["pclass"] == 1)
]

print ("12. Women in 1st class survival rate:",
    women_first["survived"].mean())


# 13. Survival rate for men in 3rd  class
men_third = df[
    (df["sex"] == "male") &
    (df["pclass"] == 3)
]

print ("13. Men in 3rd class survival rate:",
    men_third["survived"].mean())


# 14. Average family size
df["family_size"] = df["sibsp"] + df["parch"]

print("14. Average family size:",
    df["family_size"].mean())


# 15. Survival rate: alone vs with family
print("15. Survival rate by travelling alone:")

print(df.groupby("alone")["survived"].mean())



```

1. Filter vs groupby:

   For one specific group, **filtering is simpler and faster to understand**. `groupby()` is better when comparing **multiple groups**.

2. Definition of alone:

   A passenger is travelling alone if **`sibsp + parch == 0`**.

3. Most surprising result:

   The survival difference between **women and men** is surprising. It shows that **gender strongly affected survival**, with women having a much higher survival rate.


## Part B — handle missing age three ways (+ reflection)

## TASK 02 Three strategies for missing age

```python
import pandas as pd


#Load dataset
df = pd.read_csv("titanic (1).csv")


# Check missing age values

missing_age_count = df["age"].isna().sum()
missing_age_fraction = df["age"].isna().mean()

print("Missing age values:", missing_age_count)
print("Missing of missing age:", missing_age_fraction)


# Strategy 1 - Drop rows with missing age

dropped_age_data = df.dropna(subset=["age"])

drop_mean_age = dropped_age_data["age"].mean()
drop_std_age = dropped_age_data["age"].std()

print("\nStrategy 1 - Drop")
print("Rows remaining:",len(dropped_age_data))
print("Mean age:", drop_mean_age)
print("Standard deviation:", drop_std_age)


# Strategy 2 - Fill missing age with overall mean


mean_age = df["age"].mean()

mean_filled_data = df.copy()
mean_filled_data["age"] = mean_filled_data["age"].fillna(mean_age)

mean_fill_mean_age = mean_filled_data["age"].mean()
mean_fill_std_age = mean_filled_data["age"].std()

print("\nStrategy 2 - Mean Fill")
print("Mean age:", mean_fill_mean_age)
print("Standard deviation:", mean_fill_std_age)



# Strategy 3 - Fill using class + sex median

group_filled_data = df.copy()

group_filled_data["age"] = (
    group_filled_data
    .groupby(["pclass", "sex"])["age"]
    .transform(lambda age_group: age_group.fillna(age_group.median()))
)

group_fill_mean_age = group_filled_data["age"].mean()
group_fill_std_age = group_filled_data["age"].std()

print("\nStrategy 3 - Group Fill")
print("Mean age:", group_fill_mean_age)
print("Standard deviation:", group_fill_std_age)



# Compare all three strategies

comparison = pd.DataFrame({
    "Strategy": ["Drop", "Mean Fill", "Group Fill"],
    "Mean Age": [
        drop_mean_age,
        mean_fill_mean_age,
        group_fill_mean_age
    ],
    "Standard Deviation": [
        drop_std_age,
        mean_fill_std_age,
        group_fill_std_age
    ]

})


print("\nComparison:")
print(comparison)



```

1. Why does single-value filling shrink standard deviation?

   Because all missing values become the **same value**, usually near the mean. This reduces the overall variation in the data.

2. When is dropping rows wrong or fine?

   * **Wrong:** When many rows have missing values or the missing data is important.
   * **Fine:** When only a few rows are missing and removing them does not bias the analysis.

3. Can imputation change conclusions? Is it dangerous?
   **Yes.** Imputation can change averages, variation, and statistical results. It can be dangerous if the chosen method creates a false pattern or hides real differences.


## TASK 03 Reflection: which strategy, and why?

**1. Compare the three missing-age strategies.**

Dropping missing ages is simple, but it removes data. Mean filling keeps all rows but reduces the standard deviation because missing values are replaced with the same value. Group filling uses the median age based on passenger class and sex, so it better preserves differences between groups.

**2. Which strategy would you choose for this dataset and why?**

I would choose **group filling** because it does not lose any rows and gives more realistic ages based on class and sex. It also preserves the data structure better and may reduce bias in later analysis.

**3. Is there a single “correct” answer?**

No, the best strategy depends on the goal. For example, if only a few rows have missing ages, **dropping those rows** may be better because it avoids adding estimated values.

**4. Why is it a trust problem if 20% of the data was imputed without mentioning it?**

Because imputation changes the original data and can affect the results. If someone does not report that 20% was imputed, others may believe the results came entirely from real observations, making the analysis less trustworthy.


1. **Is imputing missing data a statistical or ethical choice?**
   It is **both**. Statistically, we choose a method that gives reliable results; ethically, we must be transparent about changing or estimating data.

2. **How would you document missing-data decisions?**
   I would clearly report **how many values were missing, which imputation method was used, why it was chosen, and how it affected the results**. This makes the analysis transparent and trustworthy.


## Part C — filter, sort, rank

## TASK 04 -Multi-condition filters and ranking

```python
import pandas as pd

# Load dataset
df = pd.read_csv("titanic (1).csv")


# 1. Female, 1st or 2nd class, and under 30
young_first_second_class_women = df[
    (df["sex"] == "female") &
    (df["pclass"].isin([1, 2])) &
    (df["age"] < 30)

]

print("Numbewr of passengers:", len(young_first_second_class_women))

print("Survival rate:",
        young_first_second_class_women["survived"].mean())



# 2. 10 oldest passenger who survived
oldest_survivors = (
    df[df["survived"] == 1]
    .sort_values("age", ascending=False)
    .head(10)
)

print("\n  10 oldest passengers who survived:")
print(oldest_survivors[["sex", "pclass", "age", "fare"]])


# 3. Top 10% vs bottom 10% fare-payers

top_fare_limit = df["fare"]. quantile(0.90)
bottom_fare_limit = df["fare"]. quantile(0.10)

top_10_percent_fare = df[df["fare"] >= top_fare_limit]
bottom_10_percent_fare = df[df["fare"] <= bottom_fare_limit]

top_fare_survival_rate = top_10_percent_fare["survived"].mean()
bottom_fare_survival_rate = bottom_10_percent_fare["survived"].mean()

print("\n3. Top 10% fare survival rate:",
    top_fare_survival_rate)

print("Bottom 10% fare survival rate:",
    bottom_fare_survival_rate)


# 4. Sort by class ascending and fare descending
sorted_by_class_and_fare = df.sort_values(
    ["pclass", "fare"],
    ascending=[True, False]
)

print("\n4. Dataset sorted by class and fare:")
print(sorted_by_class_and_fare.head())


# 5. Describe fare and age
print("\n5. Fare description:")
print(df["fare"].describe())

print("\nAge description:")
print(df["age"].describe())


# Value counts for class
print("\nPassenger count by class:")
print(df["class"].value_counts())


# Value counts for embarkation port
print("\nPassenger count by embarkation port:")
print(df["embark_town"].value_counts())
```

1. Why use `.isin([1,2])`?

   It is cleaner and shorter than writing two OR conditions like `(pclass == 1) | (pclass == 2)`.

2. Does higher fare relate to survival?

   Yes, higher fare generally relates to higher survival, but **fare may represent other factors**, such as passenger class, wealth, cabin location, and access to lifeboats.

3. Sorting vs ranking:

   **Sorting** rearranges the data in an order. **Ranking** assigns a position or rank to each passenger based on a value, such as fare.
